This code is used to tabulate our Monkeytype scores

```python
names  = ['Commander', 'BigIth', 'Jefe', 'Climber', 'Shire', 'LittleIth', 'Macrus']
fingers = [(94,1), (104,0.99), (106,1), (107,0.98), (40,1), (98,0.96), (None,None)]
thumbs = [(44,1),(58,0.95), (60,1), (None,None), (25,0.98), (None,None), (None,None)]
```

```python
scores = {}
for name, fscore, thscore in zip(names, fingers, thumbs):
    scores[name] = {'fingers':fscore, 'thumbs':thscore} 
```

```python
import pandas as pd

scores_df = pd.DataFrame(scores).T
scores_df
```

```python
scores_df['finger_speed (wpm)'] = scores_df.fingers.apply(lambda x: x[0])
scores_df['finger acc. (%)'] = scores_df.fingers.apply(lambda x: int(x[1]*100) if x[1] is not None else None)
scores_df['adj_finger_speed (wpm)'] = scores_df.fingers.apply(lambda x: x[0]*(2*x[1]-1) if x[0] is not None else None)
scores_df['thumb_speed (wpm)'] = scores_df.thumbs.apply(lambda x: x[0])
scores_df['thumb acc. (%)'] = scores_df.thumbs.apply(lambda x: x[1]*100 if x[1] is not None else None)
scores_df['adj_thumb_speed (wpm)'] = scores_df.thumbs.apply(lambda x: x[0]*(2*x[1]-1) if x[0] is not None else None)
```

```python

scores_df.drop(columns=['fingers', 'thumbs']).sort_values(by='adj_finger_speed (wpm)', ascending=False)
```
