According to https://arxiv.org/html/2602.14487v3, you can flip coins to calculate pi! I read through the manuscript and didn't find any major problems with it.
A minor issue I see is that the limit on the sum for the expected value of $H_\tau/\tau$ isn't infinity, but k will in principal go to infinity.

Let's check it through simulation.

```python
import numpy as np

#let's generate a suitable number of pseudo-random coin flips, with heads = 1, tails = -1
seed = 6644
rng = np.random.default_rng(seed)
n_flips = 1_000_000
#generates n_flips number of PRNs
U = rng.random(n_flips)
#convert to heads and tails
flips = (np.ceil(2*U)*2-3).astype('int64')
#print(flips)
flip_indices = np.arange(n_flips)

#now let's find the index locations at which the number of heads increases the number of tails for a given run.
#To do this, we calculate the cumulative sum of flips, and record the index of every point at which the cumulative sum
#increases above its previous maximum. The difference in the indices is the stopping time tau, and the number of heads is 
#half of the tau rounded up

cume = np.cumsum(flips)
#The number of runs in which the number of heads in sequential tosses is greater than the number of tails is just the max of the cumulative sum. 
n_runs = cume.max() if cume.max()>0 else 0 #it's possible to get zero runs of mostly heads
#array to record the indices
run_indices = np.zeros(n_runs)

#can we reduce the run time by decreasing the memory used? 1 billion 64 bit integers is 8 GB!
vtype = 'int64'
if n_runs <= 2**7 - 1:
    vtype = 'int8'
elif n_runs <= 2**16-1:
    vtype = 'int16'
elif n_runs <= 2**32-1:
    vtype = 'int32'
cume = cume.astype(vtype)

#loop through the runs and record the indices where runs end
for i in range(n_runs):
    run_indices[i] =flip_indices[np.where(cume==i+1)][0]

#now we can use the run indices to find tau and H_tau
#for H_tau we just take the difference between successive array entries, and we need to include the first entry
if n_runs>0:
    #take the first value in run_indices, then append the difference in the rest of the run indices to calculate the run lengths
    tau = np.append(run_indices[0]+1,run_indices[1:]-(run_indices)[:-1])
    H_tau = np.ceil(tau/2)

#finally, we can calculate pi
print('pi is approximately {0:.5f}'.format(4*(H_tau/tau).mean()) )
```

Now let's plot the cumulative average of the ratio vs number of runs to visualize the convergence.

```python
import matplotlib.pyplot as plt
ratio = H_tau/tau

#calculate the cumulative average, and multiply by four so that it converges to pi
converging = 4*np.cumsum(ratio)/np.arange(1,len(tau)+1)
attractor = np.pi*np.ones_like(converging)

plt.plot(attractor, color = 'black', linestyle = '--')
plt.plot(converging, color = 'orange')
plt.ylim(0,1.1*converging.max())
plt.xlabel('Run Number')
plt.ylabel('Running Average of Heads Over Run Length (x4)')
plt.annotate(text='pi', xy=[-50,3.1])
plt.title('Convergence to pi of Counting Heads in Coin Flips')

plt.show()
```

# Add in tails

An intriguing idea from LinkedIn was to also analyze the same data for tails runs, to improve the convergence efficiency. Let's try adding that in, using mostly copy and paste:


```python
#To count tails, we just take the negative of flips. 
cume_T = np.cumsum(-flips)
#The number of runs in which the number of tails in sequential tosses is greater than the number of tails is just the max of the cumulative sum. 
n_runs = cume_T.max() if cume_T.max()>0 else 0 #it's possible to get zero runs of mostly heads
#array to record the indices
run_indices = np.zeros(n_runs)

#can we reduce the run time by decreasing the memory used? 1 billion 64 bit integers is 8 GB!
vtype = 'int64'
if n_runs <= 2**7 - 1:
    vtype = 'int8'
elif n_runs <= 2**16-1:
    vtype = 'int16'
elif n_runs <= 2**32-1:
    vtype = 'int32'
cume_T = cume_T.astype(vtype)

#loop through the runs and record the indices where runs end
for i in range(n_runs):
    run_indices[i] =flip_indices[np.where(cume_T==i+1)][0]

#now we can use the run indices to find tau_T and T_tau
#for T_tau we just take the difference between successive array entries, and we need to include the first entry
if n_runs>0:
    tau_T = np.append(run_indices[0]+1,run_indices[1:]-(run_indices)[:-1])
    T_tau = np.ceil(tau_T/2)

#finally, we can calculate pi
print('pi is approximately {0:.5f}'.format(4*(T_tau/tau_T).mean()) )
```

Now we can plot the convergence of counting tails

```python
import matplotlib.pyplot as plt
ratio_T = T_tau/tau_T

#calculate the cumulative average, and multiply by four so that it converges to pi
converging_T = 4*np.cumsum(ratio_T)/np.arange(1,len(tau_T)+1)

plt.plot(attractor, color = 'black', linestyle = '--')
plt.plot(converging_T, color = 'orange')
plt.ylim(0,1.1*converging_T.max())
plt.xlabel('Run Number')
plt.ylabel('Running Average of Tails Over Run Length (x4)')
plt.annotate(text='pi', xy=[-50,3.1])
plt.title('Convergence to pi of Counting Tails in Coin Flips')

plt.show()
```

There were far fewer tails runs. Intriguingly, though, the error in tails seems to be negatively correlated with that in heads. 

We can try adding them together, run by run . 

```python
l = min(len(converging), len(converging_T))
added_converge = 0.5*(converging[:l]+converging_T[:l])
print(added_converge[-1])

plt.plot(attractor, color = 'black', linestyle = '--')
plt.plot(added_converge, color = 'orange')
plt.ylim(0,1.1*added_converge.max())
plt.xlabel('Run Number')
plt.ylabel('Running Average of Heads+Tails Over Run Lengths (x4)')
plt.annotate(text='pi', xy=[-50,3.1])
plt.title('Convergence to pi of Counting Heads and Tails in Coin Flips')

plt.show()
```

let's also try averaging all of them. 

```python
HT_tau = np.concat([H_tau, T_tau])
tau_HT = np.concat([tau, tau_T])

ratio_HT = HT_tau/tau_HT

#calculate the cumulative average, and multiply by four so that it converges to pi
converging_HT = 4*np.cumsum(ratio_HT)/np.arange(1,len(tau_HT)+1)
attractor = np.pi*np.ones_like(converging_HT)

plt.plot(attractor, color = 'black', linestyle = '--')
plt.plot(converging_HT, color = 'orange')
plt.ylim(0,1.1*converging_HT.max())
plt.xlabel('Run Number')
plt.ylabel('Running Average of (Heads,Tails) Over Run Length (x4)')
plt.annotate(text='pi', xy=[-50,3.1])
plt.title('Convergence to pi of Counting (Heads, Tails) in Coin Flips')

plt.show()
```
